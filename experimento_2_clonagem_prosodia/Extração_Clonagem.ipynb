```python
import torch
from TTS.api import TTS
from IPython.display import Audio, display

# Direciona o processamento para a GPU do Colab
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Processamento alocado em: {device}")

# Carrega o modelo multilingual XTTS v2
print("Carregando o modelo XTTS v2... Aguarde o download.")
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)
print("Motor TTS pronto e estático na memória.")
```

```python
# O "pensamento" simulado que veio do seu RAG/LLM
texto_agente = "A arquitetura está rodando perfeitamente. O vetor de referência foi extraído e o timbre foi adaptado sem necessidade de re-treinamento da rede neural base."

arquivo_referencia = "referencia.wav"  # O áudio que você subiu
arquivo_saida = "agente_clonado.wav"

print(f"Extraindo embeddings de {arquivo_referencia} e gerando síntese...")

# O modelo cruza a inteligência de pronúncia (pt) com a identidade do áudio
tts.tts_to_file(
    text=texto_agente,
    speaker_wav=arquivo_referencia,
    language="pt",
    file_path=arquivo_saida
)

print("Síntese concluída!")
display(Audio(arquivo_saida, autoplay=True))
```
