````python
from pathlib import Path
from langchain_unstructured import UnstructuredLoader

base_path = Path("/home/yuri/Projetos/Langchain_2.0/documentos")

files = [
    str(p)
    for p in base_path.iterdir()
    if p.is_file() and p.suffix.lower() in {".pdf", ".docx", ".txt"}
]

loader = UnstructuredLoader(
    files,
    chunking_strategy="basic",
    max_characters=1500,
    include_orig_elements=False,
    languages=["por"]
)

docs = loader.load()

len(docs)
````

````python
from langchain_unstructured import UnstructuredLoader
from unstructured.cleaners.core import (
    clean_extra_whitespace,
    replace_unicode_quotes,
)

loader = UnstructuredLoader(
    files,
    chunking_strategy="basic",
    max_characters=1500,
    include_orig_elements=False,
    languages=["por"],
    post_processors=[
        clean_extra_whitespace,      # remove quebras e espaços inúteis
        replace_unicode_quotes,      # normaliza aspas “ ” ‘ ’
    ],
)

docs = loader.load()
````

````python
#Testar o conteúdo do primeiro documento carregado
docs[0].page_content[:100]
````

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("BAAI/bge-m3")
````

````python
#Verificar quantidade de tokens em cada documento

def count_tokens(text):
    return len(tokenizer.encode(text))

for i, doc in enumerate(docs[:5]):
    print(i, count_tokens(doc.page_content))
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter.from_huggingface_tokenizer(
    tokenizer=tokenizer,
    chunk_size=256,
    chunk_overlap=32,
)
split_docs = text_splitter.split_documents(docs)
len(split_docs)
````

````python
from langchain_ollama.llms import OllamaLLM
from langchain_ollama import OllamaEmbeddings

modelo_1 = OllamaLLM(model="gemma3:4b")
modelo_2 = OllamaLLM(model="gemma3:1b")
modelo_embeddings = OllamaEmbeddings(model="bge-m3:567m")
````

````python
import faiss
from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore

# dimensão do embedding
embedding_dim = len(modelo_embeddings.embed_query("teste"))

# índice FAISS (L2)
index = faiss.IndexFlatL2(embedding_dim)

# vector store explícito
vector_store = FAISS(
    embedding_function=modelo_embeddings,
    index=index,
    docstore=InMemoryDocstore(),
    index_to_docstore_id={}
)

# adiciona documentos
vector_store.add_documents(split_docs)

# retriever
retriever = vector_store.as_retriever(
    search_kwargs={"k": 5}
)
````

````python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Responda usando exclusivamente os conteúdo fornecido. \n\nContexto:\n{contexto}"),
        ("human", "{query}")
    ]
)
````

````python
from langchain_core.output_parsers import StrOutputParser

cadeia = prompt | modelo_1 | StrOutputParser()
````

````python
from langchain_classic.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import json
import re

# 1. PROMPT REVISADO - Mais enfático sobre o formato
qa_generation_template = """
Com base EXCLUSIVAMENTE no trecho de documento abaixo, crie UM par útil de Pergunta e Resposta.
A pergunta deve ser específica e diretamente respondível pelo texto.
A resposta deve ser uma extração concisa do texto.

IMPORTANTE: Responda APENAS com um objeto JSON válido, sem nenhum texto adicional, sem Markdown, sem ```json.

FORMATO DE SAÍDA (JSON PURO):
{{
    "pergunta": "sua pergunta aqui",
    "resposta": "a resposta extraída do texto aqui"
}}

Trecho do Documento:
{document_text}

JSON:
"""

prompt = PromptTemplate.from_template(qa_generation_template)
qa_generation_chain = prompt | modelo_1 | StrOutputParser()

# 2. FUNÇÃO para limpar e parsear JSON
def parse_qa_response(response_text):
    """Limpa blocos de código Markdown e tenta parsear JSON."""

    # Remove blocos ```json e ``` (com espaços)
    cleaned = re.sub(r'^```json\s*|\s*```$', '', response_text.strip(), flags=re.MULTILINE)

    # Também remove possíveis "```" no início/fim
    cleaned = cleaned.strip('`')

    # Tenta parsear
    try:
        return json.loads(cleaned)
    except json.JSONDecodeError:
        # Tenta encontrar conteúdo JSON dentro do texto
        match = re.search(r'\{.*\}', cleaned, re.DOTALL)
        if match:
            try:
                return json.loads(match.group())
            except:
                pass
        raise  # Re-lança o erro se não conseguir

# 3. EXECUÇÃO corrigida
print("Gerando pares de Pergunta-Resposta...")
perguntas_respostas = []

for i, doc in enumerate(split_docs[:15]):  # Aumentei para 15 para testar
    try:
        print(f"Processando documento {i+1}...")

        # Invoca a cadeia (limita o texto para evitar tokens excessivos)
        qa_text = qa_generation_chain.invoke({
            "document_text": doc.page_content[:800]  # Reduzi para 800 caracteres
        })

        # DEBUG: Mostra o que o modelo retornou (primeiros 150 chars)
        print(f"   Raw: {qa_text[:150]}...")

        # Parseia com a função de limpeza
        qa_dict = parse_qa_response(qa_text)

        # Valida se tem os campos esperados
        if "pergunta" in qa_dict and "resposta" in qa_dict:
            qa_dict["contexto_original"] = doc.page_content[:200] + "..."
            perguntas_respostas.append(qa_dict)
            print(f"   ✅ Sucesso!")
        else:
            print(f"   ⚠️  Estrutura inválida: {qa_dict}")

    except json.JSONDecodeError as e:
        print(f"   ❌ JSON inválido após limpeza: {e}")
        print(f"   Texto após limpeza: {qa_text[:100]}...")
    except Exception as e:
        print(f"   ❌ Erro inesperado: {type(e).__name__}: {e}")

# 4. RESULTADOS
print(f"\n{'='*60}")
print(f"✅ Geração concluída! {len(perguntas_respostas)} pares QA criados com sucesso.")
print(f"{'='*60}\n")

if perguntas_respostas:
    # Salva em arquivo para uso futuro
    with open("pares_qa_gerados.json", "w", encoding="utf-8") as f:
        json.dump(perguntas_respostas, f, ensure_ascii=False, indent=2)

    print("📁 Pares salvos em 'pares_qa_gerados.json'")

    # Mostra exemplos
    print("\n📄 EXEMPLOS GERADOS:")
    for i, par in enumerate(perguntas_respostas[:3], 1):
        print(f"\n--- PAR {i} ---")
        print(f"❓ Pergunta: {par['pergunta']}")
        print(f"✅ Resposta: {par['resposta']}")
        if len(par.get('resposta', '')) > 150:
            print(f"   (Resposta truncada: {par['resposta'][:150]}...)")
else:
    print("❌ Nenhum par foi gerado. Verifique os logs acima.")
````

````python
# 1. IMPORTS e Configuração
from ragas import evaluate
from ragas.metrics.collections import faithfulness, answer_relevancy, context_recall, context_precision
from datasets import Dataset
import json
import pandas as pd

# 2. CARREGAR o dataset gerado
with open("pares_qa_gerados.json", "r", encoding="utf-8") as f:
    pares_qa = json.load(f)

# 3. FUNÇÃO para capturar contexto USADO pela sua RAG
def obter_resposta_e_contexto(pergunta):
    """
    Invoca sua cadeia RAG e captura a resposta + os documentos de contexto usados.
    """
    # A) Execute sua RAG para obter a resposta final
    resposta_final = rag_chain.invoke(pergunta)  # SUA CADEIA RAG PRINCIPAL

    # B) IMPORTANTE: Capture os documentos de contexto que foram recuperados
    #    Isso depende de como sua 'rag_chain' está estruturada.
    #    Se você tem uma 'rag_chain' que usa 'RunnablePassthrough',
    #    você pode precisar acessar um dicionário intermediário.
    #    Vamos usar uma abordagem mais direta: usar o retriever separadamente.
    #    Substitua 'seu_retriever' pela sua variável de retriever (ex: multi_query_retriever)
    documentos_recuperados = retriever.invoke(pergunta)
    lista_de_contextos = [doc.page_content for doc in documentos_recuperados]

    return resposta_final, lista_de_contextos

# 4. PREPARAR os dados para o RAGAS (com nomes de coluna CORRETOS)
print("Executando perguntas na cadeia RAG e coletando dados...")
dados_para_avaliacao = []

for i, par in enumerate(pares_qa):
    print(f"  Processando par {i+1}/{len(pares_qa)}...")

    pergunta = par["pergunta"]
    resposta_referencia = par["resposta"]  # Ground Truth

    # Obter resposta da RAG e contextos usados
    resposta_rag, contextos_usados = obter_resposta_e_contexto(pergunta)

    # Estruturar no formato EXATO que o RAGAS espera
    dados_para_avaliacao.append({
        "question": pergunta,                # string
        "answer": resposta_rag,              # string (resposta do seu RAG)
        "contexts": contextos_usados,        # lista de strings (contextos usados)
        "reference": resposta_referencia     # string (resposta de referência) - NOME CORRETO
    })

# 5. CRIAR o Dataset do RAGAS
#    O RAGAS espera as colunas: 'question', 'answer', 'contexts', 'reference'
dataset_ragas = Dataset.from_list(dados_para_avaliacao)

# Verifique se as colunas estão corretas
print(f"\n✅ Dataset criado com {len(dataset_ragas)} exemplos.")
print(f"   Colunas disponíveis: {dataset_ragas.column_names}")

# 6. EXECUTAR a avaliação
print("\n" + "="*60)
print("Iniciando avaliação RAGAS (isso pode levar alguns minutos)...")
print("="*60)

resultado = evaluate(
    dataset=dataset_ragas,
    metrics=[
        faithfulness,       # A resposta é fiel aos contextos?
        answer_relevancy,  # A resposta é relevante para a pergunta?
        context_recall,    # Os contextos cobrem toda a informação da referência?
        context_precision  # Os contextos recuperados são precisos?
    ]
)

# 7. ANALISAR os resultados
df_resultados = resultado.to_pandas()

print("\n📊 RESULTADOS DA AVALIAÇÃO - MÉDIAS:")
print("="*50)
print(f"Faithfulness:      {df_resultados['faithfulness'].mean():.2%}")
print(f"Answer Relevancy:  {df_resultados['answer_relevancy'].mean():.2%}")
print(f"Context Recall:    {df_resultados['context_recall'].mean():.2%}")
print(f"Context Precision: {df_resultados['context_precision'].mean():.2%}")
print("="*50)

# 8. (OPCIONAL) Análise detalhada dos piores casos
print("\n🔍 ANÁLISE DOS CASOS COM BAIXA FAITHFULNESS (possíveis alucinações):")
df_baixa_fidelidade = df_resultados[df_resultados['faithfulness'] < 0.7]

if not df_baixa_fidelidade.empty:
    for idx, row in df_baixa_fidelidade.head(3).iterrows():
        print(f"\n--- Índice {idx} | Faithfulness: {row['faithfulness']:.2%} ---")
        print(f"Pergunta: {row['question']}")
        print(f"Referência (resposta correta): {row['reference'][:100]}...")
        print(f"Resposta da RAG: {row['answer'][:150]}...")
        print(f"Número de contextos usados: {len(row['contexts'])}")
else:
    print("✅ Nenhum caso com faithfulness baixo (<70%) encontrado!")

# 9. Salvar resultados completos
df_resultados.to_csv("resultados_completos_ragas.csv", index=False, encoding='utf-8-sig')
print("\n📁 Resultados detalhados salvos em 'resultados_completos_ragas.csv'")
````
